# EDA - Anuncios de venta de coches (vehicles_us.csv)

Análisis exploratorio básico del conjunto de datos de anuncios de venta de coches usados en Estados Unidos, como paso previo al desarrollo de la aplicación web con Streamlit.

```python
import pandas as pd
import plotly.express as px
```

## Cargar los datos

```python
car_data = pd.read_csv('../vehicles_us.csv')
car_data.head()
```

```python
car_data.info()
```

El dataset contiene 51,525 anuncios y 13 columnas. Hay valores ausentes en `model_year`, `cylinders`, `odometer`, `paint_color` e `is_4wd` — algo esperable en anuncios publicados por usuarios, donde no siempre se completa toda la información del vehículo.

```python
car_data.describe()
```

## Histograma: distribución del kilometraje (odómetro)

```python
fig = px.histogram(car_data, x='odometer', nbins=50, title='Distribución del kilometraje')
fig.show()
```

La mayoría de los vehículos tiene entre 50,000 y 150,000 millas recorridas, con una cola larga hacia la derecha (algunos vehículos con kilometrajes muy altos, poco frecuentes).

## Histograma: distribución del precio

```python
fig = px.histogram(car_data, x='price', nbins=50, title='Distribución del precio')
fig.show()
```

## Gráfico de dispersión: precio vs. kilometraje

```python
fig = px.scatter(car_data, x='odometer', y='price', title='Precio vs. kilometraje')
fig.show()
```

Se observa una relación negativa entre el kilometraje y el precio: en general, a mayor kilometraje, menor precio de venta — un patrón esperado para vehículos usados, ya que el desgaste reduce su valor de reventa.

## Conteo de anuncios por tipo de vehículo

```python
fig = px.bar(car_data['type'].value_counts().reset_index(), x='type', y='count',
             title='Número de anuncios por tipo de vehículo')
fig.show()
```

**Conclusión:** el dataset está en buenas condiciones para construir la aplicación web. El histograma del odómetro y el gráfico de dispersión precio vs. kilometraje (implementados aquí como prueba) se incorporarán a la aplicación Streamlit en `app.py`.
